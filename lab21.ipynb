#Task 3:
Write a program to validate a password based on the following criteria:
* Minimum 8 characters
* At least one uppercase letter
* At least one lowercase letter
* At least one number
* At least one special character (e.g., @, #, !, etc.)
* Use a while loop to repeatedly ask for a password until a valid one is
entered. Use decision-making to give feedback to the user about which criteria
were not met.


```python
password = input("Enter your password: ")

feedback = []
is_valid = True

# Criteria 1: Minimum 8 characters
if len(password) < 8:
    feedback.append("  - Password must be at least 8 characters long.")
    is_valid = False

# Criteria 2: At least one uppercase letter
has_uppercase = False
for char in password:
    if char.isupper():
        has_uppercase = True
        break
if not has_uppercase:
    feedback.append("  - Password must contain at least one uppercase letter.")
    is_valid = False

# Criteria 3: At least one lowercase letter
has_lowercase = False
for char in password:
    if char.islower():
        has_lowercase = True
        break
if not has_lowercase:
    feedback.append("  - Password must contain at least one lowercase letter.")
    is_valid = False

# Criteria 4: At least one number
has_digit = False
for char in password:
    if char.isdigit():
        has_digit = True
        break
if not has_digit:
    feedback.append("  - Password must contain at least one number.")
    is_valid = False

# Criteria 5: At least one special character
special_characters_set = "!@#$%^&*()_+-=[]{};:'\"|,.<>/?`~"
has_special_char = False
for char in password:
    if char in special_characters_set:
        has_special_char = True
        break
if not has_special_char:
    feedback.append("  - Password must contain at least one special character (e.g., !@#$%^&*).")
    is_valid = False

if is_valid:
    print("\nPassword is valid! Access granted.")
else:
    print("\nPassword does not meet the following criteria:")
    for msg in feedback:
        print(msg)
    print("Please try again.\n")
```

#Task 4:
Create a complex, nested dictionary representing IIIT Ranchi’s
organizational structure.
(Visit:https://iiitranchi.ac.in/organizational_structure.aspx).
Write a Python program that accepts the name of an Assistant Professor and
iterates through the dictionary to search for that professor. If the name is found,
print the department they belong to. If the name is not found at the current level,continue to the next level of the dictionary. If the name is not found in the entire
structure, display a message indicating that the Assistant Professor was not found.


```python
iiit_ranchi_org_structure = {
    "Computer Science and Engineering": {
        "Assistant Professors": [
            "Dr. Ashok Kumar Singh",
            "Dr. Sumit Kumar",
            "Dr. Manasi Nalawade",
            "Dr. Santosh Kumar"
        ]
    },
    "Electronics and Communication Engineering": {
        "Assistant Professors": [
            "Dr. Priyanka Kumari",
            "Dr. Vijay Nath",
            "Dr. Vinay Kumar Singh"
        ]
    },
    "Information Technology": {
        "Assistant Professors": [
            "Dr. Sudip Kumar Saha",
            "Dr. Sudipta Roy",
            "Dr. Vijay Kumar"
        ]
    }

}

print("Organizational structure dictionary created. Please ensure it is accurately populated with data from the IIIT Ranchi website.")
```

```python
def find_professor_department(org_structure, professor_name):
    """
    Searches the organizational structure for a given professor's name
    and returns their department.
    """
    for department, details in org_structure.items():
        if "Assistant Professors" in details:
            if professor_name in details["Assistant Professors"]:
                return department
    return "Not Found"

print("Search function 'find_professor_department' defined. Ready to search.")
```

```python
professor_to_find = input("Enter the name of the Assistant Professor to search for: ")
department_found = find_professor_department(iiit_ranchi_org_structure, professor_to_find)

if department_found != "Not Found":
    print(f"{professor_to_find} works in the {department_found} department.")
else:
    print(f"Assistant Professor '{professor_to_find}' was not found in the organizational structure.")
```

#Task 5:
Create a list of sales data for the past 12 months (e.g., sales figures in
thousands). Calculate the total sales, average sales, and the month with the highest
sales using loops. Use matplotlib to plot the monthly sales and highlight the
month with the highest sales.
Example: # Sales data (in thousands)
sales_data = [25, 32, 29, 35, 41, 40, 45, 38, 50, 55, 60, 70]

```python
sales_data = [25, 32, 29, 35, 41, 40, 45, 38, 50, 55, 60, 70]
print(sales_data)

total_sales = 0
highest_sales = 0
highest_sales_month = 0

for i, sales in enumerate(sales_data):
    total_sales += sales
    if sales > highest_sales:
        highest_sales = sales
        highest_sales_month = i + 1

average_sales = total_sales / len(sales_data)

print(f"Total Sales: {total_sales} (in thousands)")
print(f"Average Sales: {average_sales:.2f} (in thousands)")
print(f"Highest Sales: {highest_sales} (in thousands) in Month {highest_sales_month}")


#plot for the SALES details
import matplotlib.pyplot as plt
print("Matplotlib imported successfully.")

months = range(1, len(sales_data) + 1)

plt.figure(figsize=(10, 6))
plt.plot(months, sales_data, marker='o', linestyle='-', color='skyblue', label='Monthly Sales')

# Highlight the highest sales month
plt.plot(highest_sales_month, highest_sales, marker='X', markersize=10, color='red', label=f'Highest Sales (Month {highest_sales_month})')

plt.xlabel('Month')
plt.ylabel('Sales (in thousands)')
plt.title('Monthly Sales Data and Highest Sales Month')
plt.xticks(months)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
