iron man recorded 7 hit out of 10 short both in integer convert them into float and then calculate the accuracy%

```python
hits=7
short=10
hit_f=float(hits)
short_f=float(short)
accuracy =(hit_f*100)/(short_f)
print(f"accuracy={accuracy: .2f}%")
```
