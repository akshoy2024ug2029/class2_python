# Task 1:
Create a text file students.txt and write some student names and
grades to it. Write a program to read the file and display the contents. Implement
error handling for cases like file not found and permission errors.

```python
with open('students.txt', 'w') as file:
    file.write('Rahul, 85\n')
    file.write('Rohit, 92\n')
    file.write('Abhishek, 78\n')
print('students.txt created and data written successfully.')

try:
    with open('students.txt', 'r') as file:
        content = file.read()
        print('File content:')
        print(content)
except FileNotFoundError:
    print("Error: The file 'students.txt' was not found.")
except PermissionError:
    print("Error: Permission denied to access 'students.txt'.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
