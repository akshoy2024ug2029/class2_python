### Program to compute student marks and classify into grades

```python
# Example scores for a single student
student_name = "Alice"
scores = [85, 90, 78, 92] # List of scores for the student

if not scores:
    print(f"Student {student_name} has no scores.")
else:
    total_score = sum(scores)
    average_score = total_score / len(scores)

    # Grade classification logic
    if average_score >= 90:
        grade = 'A'
    elif average_score >= 80:
        grade = 'B'
    elif average_score >= 70:
        grade = 'C'
    elif average_score >= 60:
        grade = 'D'
    else:
        grade = 'F'

    print(f"Student: {student_name}")
    print(f"  Scores: {scores}")
    print(f"  Total Score: {total_score}")
    print(f"  Average Score: {average_score:.2f}")
    print(f"  Grade: {grade}")
```
