1. Define a function rectangle area perimeter that takes the length and width as input parameters. Return both the area and perimeter of the rectangle from the function. Call the function with different sets of length and width values.



```python
def rectangle_area_perimeter(length, width):#function to calculate area and perimeter of a rectangle
  area = length * width #formulae for area of rectangle
  perimeter = 2 * (length + width)#formulae for perimeter of rectangle
  return area, perimeter #returns area and perimeter

#Examples of some length and breadth of rectangles
print(rectangle_area_perimeter(5, 10))
print(rectangle_area_perimeter(7, 7))
print(rectangle_area_perimeter(2.5, 4))
```

2.  Define a function list summary that accepts a list of integers. Return the sum, average, and maximum value of

the list as a tuple. Test the function with a sample list.



```python
def List_summary(list):#function to calculate sum,average,maximum
  sum=0
  for i in list:#to iterare throughout the list
    sum+=i
    maximum=max(list)#provides the maximum
    average=sum/len(list)#calculates the average
  return sum,average,maximum


list=[2,7,9,8,10,11,25]
print("(sum,average,maximum)")
print(List_summary(list))#calls the functions and prints the returned value

```

3. Import the datetime module. Define a function get day_of week that takes a date string in YYYY-MM- DD format. Use date datetime.strptime() to convert the date string into a datetime object and then use strftime(%A) to get the day of the week. Test the function with various dates

```python
import datetime #imports datetime module
def get_day_of_week(date_string): #Function to find the day of the week from the given date
  date=datetime.datetime.strptime(date_string,"%Y-%m-%d") #converts the date string into the real time date
  day_of_week=date.strftime("%A") #finds the day of the week
  return day_of_week #finally returns the day of the week

date_string="2025-09-19"
print(get_day_of_week(date_string)) #calls the function and prints the day of the week
```

4. Define a function highest grade that accepts a dictionary with student names es keys and their grades values. Use the max() function to find the student with the e to highest grade. Test th: function with a sample dictionary.

```python
def highest_grade(dictionary):
  highest_grade=max(dictionary.values())
  for key,value in dictionary.items():
    if value==highest_grade:
      return key,value

st_marks={"akhil":25,
         "ananya":26,
         "gaurav":15,
          "akshay":21,
          "rahul":24}
print(highest_grade(st_marks))
```

5. Import the random and math modules. Generate a random number between 1 and 100 using random.randint(). Use the math module to check

if the number is prime. (A prime number is a number greater than 1 that has no divisors other than 1 and itself.) Print whether the random number

is prime or not.

```python
import random#imports the random library
import math#imports the math library for calculation
generate=random.randint(1,100)#generates a random integer between the given range
#finding if the generated random integer is prime or not
is_prime = True
if generate <= 1:
    is_prime = False
else:
    for i in range(2, int(math.sqrt(generate)) + 1):
        if generate % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{generate} is prime") #prints if number is prime
else:
    print(f"{generate} is not prime")#prints if number is not prime
```
